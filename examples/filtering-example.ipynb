```python
%load_ext autoreload
%autoreload 2

import importlib
import sys

sys.path.insert(1, '..')

import numpy as np
import matplotlib.pyplot as plt

from qcd_tools import filtering

plt.rcParams["figure.figsize"] = (6, 3)
plt.rcParams["figure.dpi"] = 100
plt.rcParams["savefig.dpi"] = 100
plt.rcParams["legend.loc"] = 'upper right'
plt.rc('font', size=8)
plt.rc('axes', titlesize=12)
plt.rc('axes', labelsize=12)
plt.rc('xtick', labelsize=8)
plt.rc('ytick', labelsize=8)
plt.rc('legend', fontsize=8)
plt.rc('figure', titlesize=8)
```

```python
poles = np.loadtxt("data/padefit-example-poles-re.dat") + 1.0j * np.loadtxt("data/padefit-example-poles-im.dat")

plt.figure()
plt.scatter(np.real(poles),np.imag(poles), marker='.', color='r')
plt.xlabel('$Re[x]$')
plt.ylabel('$Im[x]$')
plt.xlim([-0.5,6.5])
plt.ylim([-0.5,6.5])
plt.title("$Singularities$")

plt.show()
```

```python
sings = filtering.apply_filter(poles, (0.0,np.inf,0.0,np.inf))

plt.figure()
plt.scatter(np.real(sings),np.imag(sings), marker='.', color='r')
plt.xlabel('$Re[x]$')
plt.ylabel('$Im[x]$')
plt.xlim([-0.5,6.5])
plt.ylim([-0.5,6.5])
plt.title("$Singularities$")

plt.show()
```

```python
clusterR = filtering.get_cluster(sings, which_cluster='closest_to_real')
clusterI = filtering.get_cluster(sings, which_cluster='closest_to_imaginary')

plt.figure()
plt.scatter(np.real(clusterI),np.imag(clusterI), marker='.', color='r')
plt.scatter(np.real(clusterR),np.imag(clusterR), marker='.', color='b')
plt.xlabel('$Re[x]$')
plt.ylabel('$Im[x]$')
plt.xlim([-0.5,6.5])
plt.ylim([-0.5,6.5])
plt.title("$Singularities$")

plt.show()
```

```python
clusterR = filtering.get_cluster(sings, which_cluster='closest_to_real', eps=0.5)
clusterI = filtering.get_cluster(sings, which_cluster='closest_to_imaginary', eps=0.5)

plt.figure()
plt.scatter(np.real(clusterI),np.imag(clusterI), marker='.', color='r')
plt.scatter(np.real(clusterR),np.imag(clusterR), marker='.', color='b')
plt.xlabel('$Re[x]$')
plt.ylabel('$Im[x]$')
plt.xlim([-0.5,6.5])
plt.ylim([-0.5,6.5])
plt.title("$Singularities$")

plt.show()
```
